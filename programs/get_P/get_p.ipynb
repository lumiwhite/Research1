```python
def get_prime_factors(n):
    factors = []
    d = 2
    while d * d <= n:
        while (n % d) == 0:
            factors.append(d)
            n //= d
        d += 1
    if n > 1:
        factors.append(n)
    return factors

LIMIT = 768

two_distinct_prime_factors = []
for i in range(2, LIMIT):
    factors = get_prime_factors(i)
    unique_factors = set(factors)
    if len(unique_factors) == 2:
        two_distinct_prime_factors.append(i)

print(f"【解釈2】2種類の素因数を持つ数 (p^a * q^b) の数: {len(two_distinct_prime_factors)} 個")
print(two_distinct_prime_factors[::-1])
```
