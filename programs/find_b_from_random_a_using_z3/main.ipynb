```python
import sys
import os
from datetime import datetime

log_dir = 'log'
os.makedirs(log_dir, exist_ok=True)

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
log_filename = os.path.join(log_dir, f"{timestamp}_search_progress.log")
print(f"探索の進捗を {log_filename} に記録します。")
original_stdout = sys.stdout

# buffering=1 を指定することで「行バッファリング（改行ごとにファイルへ書き込む）」を有効にする
f = open(log_filename, 'w', encoding='utf-8', buffering=1)

try:
    sys.stdout = f  # 出力先をファイルに変更
    
    # 動作確認のため、最初の1行を確実に出力
    print(f"{timestamp} : 探索を開始...", flush=True) 
    
    cycles = gen_cycles(6)
    h_x, h_z = gen_h_xz()
    
    # 探索のメイン処理
    a_vec, b_list = optimized_find_b_fast(cycles, h_x, h_z, 10)
    
finally:
    # tryブロック内でエラーや強制停止(KeyboardInterrupt)が起きても、
    # 必ずここが実行され、Jupyterの出力を元に戻し、ファイルを安全に閉じる
    sys.stdout = original_stdout
    f.close()

print(f"探索が完了し、ログへの書き込みを終了した。保存先: {log_filename}")
```
