```python
import math
```

```python
def solve_linear_congruence(A, B, M):
    """
    一次合同式 Ax = B (mod M) を解き、[0, M-1] の範囲の全解をリストで返す。
    解が存在しない場合は空リストを返す。
    """
    g = math.gcd(A, M)
    if B % g != 0:
        return []
    
    # 式を簡約化: (A/g)x = (B/g) (mod M/g)
    A0, B0, M0 = A // g, B // g, M // g
    # 簡約後の係数は互いに素なので逆元が存在する
    try:
        x0 = (B0 * pow(A0, -1, M0)) % M0
        return [(x0 + i * M0) % M for i in range(g)]
    except ValueError:
        return []
```

```python
def print_apm_centers(P, a_list, b_list):
    """12個のAPMについて中心座標を出力する"""
    labels = [f"f{i}" for i in range(6)] + [f"g{i}" for i in range(6)]
    
    print(f"{'関数':<5} | {'a':<5} | {'b':<5} | {'中心座標 c (不動点)':<30}")
    print("-" * 65)
    
    for i in range(12):
        a, b = a_list[i], b_list[i]
        # 不動点条件: (a-1)c = -b (mod P)
        A = (a - 1) % P
        B = (-b) % P
        
        centers = solve_linear_congruence(A, B, P)
        
        if not centers:
            center_str = "None (解なし)"
        elif len(centers) > 4:
            center_str = f"{len(centers)}個の解 (例: {centers[:3]}...)"
        else:
            center_str = str(centers)
            
        print(f"{labels[i]:<5} | {a:<5} | {b:<5} | {center_str}")
```

```python
a_vec = [763, 679, 397, 61, 697, 373, 289, 257, 625, 41, 193, 449]
b_vec = [15, 489, 18, 762, 660, 174, 48, 384, 488, 764, 160, 544]
```

```python
print_apm_centers(768, a_vec, b_vec)
```
