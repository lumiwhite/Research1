```python
def check_condition_ab(a_vec, b_vec):
    Ga, Gb = gen_g_mat(a_vec)
    b_vec_np = np.array(b_vec)
    is_all_zero = ((Ga @ b_vec_np % P) == 0).all()
    is_all_non_zero = ((Gb @ b_vec_np % P) != 0).all()
    return is_all_zero and is_all_non_zero
```

```python
def check_condition_c(a_vec, b_vec, girth):
    cycles = gen_cycles(girth)
    h_x, h_z = gen_h_xz()
    constraints = gen_c_constraints(a_vec, cycles, h_x, h_z)
    b_vec_np = np.array(b_vec)
    for constraint in constraints:
        res = np.array(constraint) @ b_vec_np % P
        if res == 0:
            return False
    return True
```

```python
def check_conditions(a_vec, b_vec, girth):
    return check_condition_ab(a_vec, b_vec) and check_condition_c(a_vec, b_vec, girth)
```
